```python
import pandas as pd
csv_file= 'uscities.csv'
zip_path = 'uscities.zip'
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    with zip_ref.open(csv_file) as file:
        uscities = pd.read_csv(file)
uscities.head(2)
```

```python
import json

# Create dictionary {city: state_name}
CITY_TO_STATE = dict(zip(uscities["city"], uscities["state_name"]))

# Create dictionary {county: state_name}
COUNTY_TO_STATE = dict(zip(uscities["county_name"], uscities["state_name"]))


# Save both dictionaries for later use (optional)
with open("city_to_state.json", "w") as f:
    json.dump(CITY_TO_STATE, f)

with open("county_to_state.json", "w") as f:
    json.dump(COUNTY_TO_STATE, f)

print("Dictionaries created successfully!")
```

```python
# Load city-to-state mapping
with open("city_to_state.json", "r") as f:
    CITY_TO_STATE = json.load(f)
# Display the first key-value pairs
first_items = list(CITY_TO_STATE.items())[:2]  
print(first_items)
```
